```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
from pathlib import Path
from matplotlib.ticker import FuncFormatter
from matplotlib.patches import Patch


# =========================
# STYLE
# =========================

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 7)


# =========================
# LOAD DATA
# =========================

DATA_PATH = "/content/drive/MyDrive/amazon-data-project/data/amazon.csv"

df = pd.read_csv(DATA_PATH, encoding="utf-8-sig")


# =========================
# CLEAN DATA
# =========================

def clean_money(series):
    return pd.to_numeric(
        series.astype(str)
              .str.replace(",", "", regex=False)
              .str.replace(r"[^\d.]", "", regex=True),
        errors="coerce"
    )


df["discounted_price"] = clean_money(df["discounted_price"])
df["actual_price"] = clean_money(df["actual_price"])

df["discount_percentage"] = pd.to_numeric(
    df["discount_percentage"]
      .astype(str)
      .str.replace("%", "", regex=False),
    errors="coerce"
)

df["rating"] = pd.to_numeric(
    df["rating"],
    errors="coerce"
)

df["rating_count"] = pd.to_numeric(
    df["rating_count"]
      .astype(str)
      .str.replace(",", "", regex=False),
    errors="coerce"
)

df["category"] = df["category"].astype(str).str.strip()

df = df.dropna(
    subset=[
        "discounted_price",
        "actual_price",
        "rating",
        "rating_count",
        "category"
    ]
)


# =========================
# INR TO USD CONVERSION
# =========================

INR_TO_USD = 0.0104311

df["discounted_price_usd"] = (
    df["discounted_price"] * INR_TO_USD
)

df["actual_price_usd"] = (
    df["actual_price"] * INR_TO_USD
)

df["discount_amount_usd"] = (
    df["actual_price_usd"] -
    df["discounted_price_usd"]
)


# =========================
# SHORT CATEGORY NAMES
# =========================

def shorten_category(category):

    short_name = str(category).split("|")[-1]

    short_name = re.sub(
        r"(?<=[a-z])(?=[A-Z])",
        " ",
        short_name
    )

    short_name = re.sub(
        r"(?<=[A-Z])(?=[A-Z][a-z])",
        " ",
        short_name
    )

    short_name = short_name.replace("&", " & ")

    short_name = re.sub(
        r"\s+",
        " ",
        short_name
    ).strip()

    return short_name


df["category_short"] = (
    df["category"]
    .apply(shorten_category)
)


print("Data cleaning completed.")
print("Prices converted from INR to USD.")
print(df.info())

display(df.head())


# =========================
# OUTPUT FOLDER
# =========================

OUTPUT_DIR = Path(
    "/content/drive/MyDrive/amazon-data-project/outputs/figures"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# USD formatter
money = FuncFormatter(
    lambda x, pos: f"${x:,.0f}"
)


# ============================================================
# GRAPH 1
# TOP 10 CATEGORIES BY AVERAGE DISCOUNT
# ============================================================

top_n = 10

avg_discount = (
    df.groupby("category")["discount_amount_usd"]
      .mean()
      .sort_values(ascending=False)
      .head(top_n)
)

avg_discount_df = avg_discount.reset_index()

avg_discount_df["category_short"] = (
    avg_discount_df["category"]
    .apply(shorten_category)
)


plt.figure(figsize=(12, 7))

ax = sns.barplot(
    data=avg_discount_df,
    x="discount_amount_usd",
    y="category_short"
)

ax.xaxis.set_major_formatter(money)

plt.title(
    "Top 10 Categories by Average Discount Amount",
    fontsize=16
)

plt.xlabel(
    "Average Discount (USD)",
    fontsize=12
)

plt.ylabel(
    "Category",
    fontsize=12
)


for i, value in enumerate(
    avg_discount_df["discount_amount_usd"]
):
    ax.text(
        value,
        i,
        f"  ${value:,.0f}",
        va="center"
    )


plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "top_categories_avg_discount.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
print("\n\n\n")


# ============================================================
# GRAPH 2
# MEAN VS MEDIAN DISCOUNT
# ============================================================

stats_cat = (
    df.groupby("category")["discount_amount_usd"]
      .agg(
          mean="mean",
          median="median",
          count="size"
      )
      .sort_values(
          "mean",
          ascending=False
      )
      .head(top_n)
      .reset_index()
)

stats_cat["category_short"] = (
    stats_cat["category"]
    .apply(shorten_category)
)


plot_df = stats_cat.melt(
    id_vars=[
        "category",
        "category_short"
    ],
    value_vars=[
        "mean",
        "median"
    ],
    var_name="metric",
    value_name="discount_amount_usd"
)

plot_df["metric"] = (
    plot_df["metric"]
    .str.title()
)


category_order = (
    stats_cat["category_short"]
    .tolist()
)


plt.figure(figsize=(12, 7))

ax = sns.barplot(
    data=plot_df,
    x="discount_amount_usd",
    y="category_short",
    hue="metric",
    order=category_order
)

ax.xaxis.set_major_formatter(money)

plt.title(
    "Top 10 Categories: Mean vs Median Discount",
    fontsize=16
)

plt.xlabel(
    "Discount Amount (USD)",
    fontsize=12
)

plt.ylabel(
    "Category",
    fontsize=12
)


colors = sns.color_palette()

legend_elements = [
    Patch(
        facecolor=colors[0],
        label="Mean"
    ),
    Patch(
        facecolor=colors[1],
        label="Median"
    )
]

ax.legend(
    handles=legend_elements,
    title="Statistic"
)


plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "mean_vs_median_discount_by_category.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
print("\n\n\n")


# ============================================================
# GRAPH 3
# DISCOUNT DISTRIBUTION BY CATEGORY
# ============================================================

top_k = 8

top_categories = (
    df.groupby("category")["discount_amount_usd"]
      .median()
      .sort_values(ascending=False)
      .head(top_k)
      .index
)

df_top = df[
    df["category"].isin(top_categories)
].copy()

df_top["category_short"] = (
    df_top["category"]
    .apply(shorten_category)
)


category_order_box = [
    shorten_category(category)
    for category in top_categories
]


plt.figure(figsize=(12, 7))

ax = sns.boxplot(
    data=df_top,
    x="discount_amount_usd",
    y="category_short",
    order=category_order_box,
    showfliers=False
)

ax.xaxis.set_major_formatter(money)

plt.title(
    "Discount Distribution Across Top Categories",
    fontsize=16
)

plt.xlabel(
    "Discount Amount (USD)",
    fontsize=12
)

plt.ylabel(
    "Category",
    fontsize=12
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "discount_boxplot_top_categories.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
print("\n\n\n")


# ============================================================
# GRAPH 4
# PRODUCT RATING VS DISCOUNT
# ============================================================

plt.figure(figsize=(11, 6))

ax = sns.regplot(
    data=df,
    x="discount_amount_usd",
    y="rating",
    scatter_kws={
        "alpha": 0.30,
        "s": 18
    },
    line_kws={
        "linewidth": 2
    },
    ci=95
)

ax.xaxis.set_major_formatter(money)

plt.title(
    "Product Rating vs Discount Amount",
    fontsize=16
)

plt.xlabel(
    "Discount Amount (USD)",
    fontsize=12
)

plt.ylabel(
    "Product Rating",
    fontsize=12
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "rating_vs_discount_regression.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
print("\n\n\n")


# ============================================================
# GRAPH 5
# DISCOUNTED PRICE DISTRIBUTION
# ============================================================

plt.figure(figsize=(11, 6))

ax = sns.histplot(
    df["discounted_price_usd"],
    bins=40,
    kde=True
)

ax.xaxis.set_major_formatter(money)

plt.title(
    "Distribution of Discounted Product Prices",
    fontsize=16
)

plt.xlabel(
    "Discounted Price (USD)",
    fontsize=12
)

plt.ylabel(
    "Number of Products",
    fontsize=12
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "discounted_price_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
print("\n\n\n")


# ============================================================
# GRAPH 6
# DISCOUNTED PRICE DISTRIBUTION - LOG SCALE
# ============================================================

plt.figure(figsize=(11, 6))

sns.histplot(
    np.log1p(df["discounted_price_usd"]),
    bins=40,
    kde=True
)

plt.title(
    "Distribution of Discounted Prices (Log Scale)",
    fontsize=16
)

plt.xlabel(
    "log1p(Discounted Price in USD)",
    fontsize=12
)

plt.ylabel(
    "Number of Products",
    fontsize=12
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "discounted_price_distribution_log.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
print("\n\n\n")


print(
    f"Charts saved in:\n{OUTPUT_DIR}"
)
```
